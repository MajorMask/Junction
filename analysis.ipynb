```python
import pandas as pd
import csv

def find_distinct_values_from_csv(file_path):
    """
    Reads a CSV file and finds all unique values for every column.

    Args:
        file_path (str): The full path to the CSV file.

    Returns:
        dict: A dictionary where keys are column names and values are 
              lists of unique values found in that column.
    """
    try:
        # 1. Read the CSV file into a Pandas DataFrame
        df = pd.read_csv(file_path)

        # 2. Initialize an empty dictionary to store results
        distinct_values = {}

        # 3. Iterate over each column in the DataFrame
        for column in df.columns:
            # The .unique() method returns an array of unique values for a Series (column)
            # We convert it to a list for easier viewing/handling
            unique_list = df[column].dropna().unique().tolist()
            distinct_values[column] = unique_list

        return distinct_values

    except FileNotFoundError:
        print(f"Error: The file '{file_path}' was not found.")
        return {}
    except pd.errors.EmptyDataError:
        print(f"Error: The file '{file_path}' is empty.")
        return {}
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        return {}

# --- Example Usage ---
# Replace 'your_file.csv' with the actual path to your CSV file
file_path = 'products.csv' 
results = find_distinct_values_from_csv(file_path)

# Print the results
if results:
    print(f"\n--- Distinct Values from {file_path} ---")
    for column, values in results.items():
        print(f"\n**{column}** ({len(values)} distinct values):")
        # Print a few values for summary, or 'values' for all
        print(values[:10]) 
        if len(values) > 10:
            print("...")
```
